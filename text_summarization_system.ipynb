```python

import re
from collections import Counter

def summarize_text(text, summary_length=3):

    # Split text into sentences
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())

    # Remove common words
    stop_words = {
        "the", "is", "are", "was", "were", "a", "an", "and",
        "or", "of", "in", "to", "for", "on", "with", "this",
        "that", "it", "as", "by", "at", "be", "from", "has",
        "have", "had", "which", "their", "they", "its"
    }

    # Calculate word frequency
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
    word_freq = Counter(
        word for word in words if word not in stop_words
    )

    # Calculate sentence scores
    sentence_scores = {}

    for sentence in sentences:
        score = 0
        for word in re.findall(r'\b[a-zA-Z]+\b', sentence.lower()):
            score += word_freq.get(word, 0)
        sentence_scores[sentence] = score

    # Select important sentences
    important_sentences = sorted(
        sentence_scores,
        key=sentence_scores.get,
        reverse=True
    )[:summary_length]

    # Maintain original sentence order
    summary = [
        sentence for sentence in sentences
        if sentence in important_sentences
    ]

    return " ".join(summary)


# Input text
text = """
Artificial Intelligence is one of the most important technologies
in the modern world. It helps machines perform tasks that normally
require human intelligence. AI is used in healthcare, education,
banking, transportation and many other fields.

Machine learning is a part of artificial intelligence. It allows
computers to learn from data and improve their performance.
AI systems can understand languages, recognize images and solve
complex problems. In the future, artificial intelligence will
continue to play an important role in human life.
"""

# Generate summary
result = summarize_text(text, 3)

print("ORIGINAL TEXT:")
print(text)

print("\nSUMMARIZED TEXT:")
print(result)
```
